```python
def profile_text(df, dataset_name):

    texts = df["text"].dropna().astype(str)

    arabic_pattern = re.compile(r"[\u0600-\u06FF]")
    english_pattern = re.compile(r"[A-Za-z]")
    number_pattern = re.compile(r"\d")

    arabic_count = 0
    english_count = 0
    mixed_count = 0
    numeric_count = 0
    multiline_count = 0

    english_words = Counter()
    special_chars = Counter()

    for text in texts:

        has_arabic = bool(arabic_pattern.search(text))
        has_english = bool(english_pattern.search(text))

        arabic_count += has_arabic
        english_count += has_english
        mixed_count += has_arabic and has_english
        numeric_count += bool(number_pattern.search(text))
        multiline_count += "\n" in text

        english_words.update(
            re.findall(r"[A-Za-z]+", text)
        )

        special_chars.update(
            char for char in text
            if not char.isalnum() and not char.isspace()
        )

    print("\n" + "=" * 60)
    print(f"TEXT PROFILE: {dataset_name}")
    print("=" * 60)

    print("Total texts:", len(texts))
    print("Arabic texts:", arabic_count)
    print("English-containing texts:", english_count)
    print("Mixed Arabic-English:", mixed_count)
    print("Texts containing numbers:", numeric_count)
    print("Multiline texts:", multiline_count)

    print("\nTop 20 English words:")
    print(english_words.most_common(20))

    print("\nTop 20 special characters:")
    print(special_chars.most_common(20))


profile_text(df_erzen, "Erzen")
profile_text(df_arabic_english, "Arabic-English")
```

```python
# ============================================================
# 1. Unicode Normalization
# ============================================================

def normalize_unicode(text: str) -> str:
    if not isinstance(text, str):
        raise TypeError("text must be a string")

    text = unicodedata.normalize("NFC", text)
    text = text.replace("\ufeff", "")
    text = text.replace("\t", " ")
    text = text.replace("\n", " ")
    text = text.replace("\r", " ")

    text = "".join(
        char for char in text
        if unicodedata.category(char) != "Cc"
    )

    text = " ".join(text.split())

    return unicodedata.normalize("NFC", text)


# ============================================================
# 2. Arabic Diacritics Removal
# ============================================================

ARABIC_DIACRITICS_PATTERN = re.compile(
    r"[\u064B-\u0652]"
)


def remove_arabic_diacritics(text: str) -> str:
    if not isinstance(text, str):
        raise TypeError("text must be a string")

    return ARABIC_DIACRITICS_PATTERN.sub("", text)


# ============================================================
# 3. Arabic Orthographic Normalization
# ============================================================

ARABIC_CHAR_TRANSLATION = str.maketrans({
    "أ": "ا",
    "إ": "ا",
    "آ": "ا",
    "ٱ": "ا",
    "ى": "ي",
    "ة": "ه",
})


def normalize_arabic_orthography(text: str) -> str:
    if not isinstance(text, str):
        raise TypeError("text must be a string")

    text = text.translate(ARABIC_CHAR_TRANSLATION)
    text = text.replace("ـ", "")

    return text


# ============================================================
# 4. Punctuation & Annotation Handling
# ============================================================

REMOVABLE_PUNCTUATION = str.maketrans({
    '"': None,
    "“": None,
    "”": None,
    "«": None,
    "»": None,
    "(": None,
    ")": None,
    "{": None,
    "}": None,
})


def normalize_punctuation(text: str) -> str:
    if not isinstance(text, str):
        raise TypeError("text must be a string")

    # Remove selected quotation marks and brackets
    text = text.translate(REMOVABLE_PUNCTUATION)

    # Remove single quotes only when they wrap the entire text
    if len(text) >= 2:
        if (text[0], text[-1]) in [
            ("'", "'"),
            ("‘", "’"),
        ]:
            text = text[1:-1]

    return " ".join(text.split())


# ============================================================
# 5. Complete Normalization Pipeline
# ============================================================

def normalize_text(text: str) -> str:
    if not isinstance(text, str):
        raise TypeError("text must be a string")

    text = normalize_unicode(text)
    text = remove_arabic_diacritics(text)
    text = normalize_arabic_orthography(text)
    text = normalize_punctuation(text)

    return text
```

```python
# ============================================================
# FINAL TEST SUITE — 50 FIXTURES
# ============================================================

normalization_fixtures = [

    # Arabic orthography
    ("أحمد", "احمد"),
    ("إسلام", "اسلام"),
    ("آدم", "ادم"),
    ("ٱلْحَمْدُ", "الحمد"),
    ("على", "علي"),
    ("مدرسة", "مدرسه"),
    ("جميلة", "جميله"),
    ("هُدى", "هدي"),
    ("شيء", "شيء"),
    ("مسؤول", "مسؤول"),

    # Arabic diacritics
    ("مُحَمَّد", "محمد"),
    ("كِتَاب", "كتاب"),
    ("مَدْرَسَة", "مدرسه"),
    ("العَرَبِيَّة", "العربيه"),
    ("شُكْرًا", "شكرا"),

    # English preservation
    ("PyTorch", "PyTorch"),
    ("TensorFlow", "TensorFlow"),
    ("Machine Learning", "Machine Learning"),
    ("Deep Learning", "Deep Learning"),
    ("SQL Server", "SQL Server"),

    # Technical tokens
    ("GPT-6", "GPT-6"),
    ("HTTP/2", "HTTP/2"),
    ("C++", "C++"),
    ("Node.js", "Node.js"),
    ("Python 3.12", "Python 3.12"),
    ("CUDA 12.4", "CUDA 12.4"),
    ("Whisper Large-v3", "Whisper Large-v3"),
    ("API", "API"),
    ("Layer Normalization", "Layer Normalization"),
    ("don't remove user's data", "don't remove user's data"),

    # Code-switching
    ("أهلاً بـ GPT-6", "اهلا ب GPT-6"),
    ("استخدم PyTorch مع CUDA", "استخدم PyTorch مع CUDA"),
    ("العَرَبِيَّة و English", "العربيه و English"),
    ("إزاي تستخدم Python؟", "ازاي تستخدم Python؟"),
    ("Layer Normalization في PyTorch", "Layer Normalization في PyTorch"),
    ("Hello يا أحمد", "Hello يا احمد"),
    ("ال+weekends", "ال+weekends"),
    ("ال+hobbies", "ال+hobbies"),
    ("Okay ماشي", "Okay ماشي"),
    ("Python 3.12 مع GPT-6", "Python 3.12 مع GPT-6"),

    # Whitespace and control characters
    ("  Hello   World  ", "Hello World"),
    ("طيب\nبتعمل إيه؟", "طيب بتعمل ايه؟"),
    ("Layer\tNormalization", "Layer Normalization"),
    ("\ufeffHello World", "Hello World"),
    ("", ""),

    # Punctuation and annotations
    ("HTTP/2 - GPT-6", "HTTP/2 - GPT-6"),
    ('"Hello World"', "Hello World"),
    ("(PyTorch 2.5)", "PyTorch 2.5"),
    ("[LAUGHTER] okay شكرا", "[LAUGHTER] okay شكرا"),
    ("Hello, world!", "Hello, world!"),
]


# ============================================================
# Run the 50 fixtures
# ============================================================

failed_tests = []

for index, (original, expected) in enumerate(
    normalization_fixtures, start=1
):
    actual = normalize_text(original)

    if actual != expected:
        failed_tests.append({
            "test": index,
            "input": original,
            "expected": expected,
            "actual": actual,
        })


# ============================================================
# Additional technical preservation tests
# ============================================================

technical_tests = [
    "PyTorch 2.5",
    "TensorFlow",
    "Python 3.12",
    "GPT-6",
    "HTTP/2",
    "C++",
    "Node.js",
    "CUDA 12.4",
    "Whisper Large-v3",
    "don't",
    "user's account",
    "[LAUGHTER]",
    "ال+weekends",
]

for original in technical_tests:
    actual = normalize_text(original)

    if actual != original:
        failed_tests.append({
            "test": "technical preservation",
            "input": original,
            "expected": original,
            "actual": actual,
        })


# ============================================================
# Idempotency tests
# ============================================================

idempotency_tests = [
    "أَحْمَد بيستخدم Python 3.12",
    '"Hello" مع PyTorch 2.5',
    "[LAUGHTER] و [HESITATION]",
    "ال+weekends و HTTP/2",
    "don't remove user's data",
]

for original in idempotency_tests:
    once = normalize_text(original)
    twice = normalize_text(once)

    if once != twice:
        failed_tests.append({
            "test": "idempotency",
            "input": original,
            "expected": once,
            "actual": twice,
        })


# ============================================================
# Final report
# ============================================================

print("=" * 55)
print("AI-DS-02 FINAL NORMALIZATION TEST REPORT")
print("=" * 55)

print("Fixtures:", len(normalization_fixtures))
print("Technical preservation tests:", len(technical_tests))
print("Idempotency tests:", len(idempotency_tests))

total_tests = (
    len(normalization_fixtures)
    + len(technical_tests)
    + len(idempotency_tests)
)

passed_tests = total_tests - len(failed_tests)

print("\nTotal tests:", total_tests)
print("Passed:", passed_tests)
print("Failed:", len(failed_tests))

if failed_tests:
    print("\nFAILED TEST DETAILS:")
    display(pd.DataFrame(failed_tests))
    raise AssertionError(
        f"{len(failed_tests)} normalization tests failed."
    )

print("\nALL NORMALIZATION TESTS PASSED.")
```

```python
def preview_normalization(df, dataset_name, sample_size=20):

    sample = df[["text"]].head(sample_size).copy()

    sample["normalized_text"] = sample["text"].apply(
        lambda text: normalize_text(text)
        if isinstance(text, str)
        else text
    )

    sample["changed"] = (
        sample["text"] != sample["normalized_text"]
    )

    print(f"\n{'=' * 70}")
    print(f"NORMALIZATION PREVIEW: {dataset_name}")
    print("=" * 70)

    display(sample)

    print("Changed texts:", int(sample["changed"].sum()))
    print("Unchanged texts:", int((~sample["changed"]).sum()))


preview_normalization(df_erzen, "Erzen")
preview_normalization(df_arabic_english, "Arabic-English")
```

```python
def process_dataset(df, dataset_name):

    cleaned_df = df.copy(deep=True)

    original_texts = cleaned_df["text"].copy()

    # Normalize text and preserve missing values
    cleaned_df["text"] = cleaned_df["text"].apply(
        lambda text: normalize_text(text)
        if isinstance(text, str)
        else text
    )

    # Correctly exclude rows where both original and
    # normalized text are missing
    changed_mask = (
        original_texts.ne(cleaned_df["text"])
        & ~(
            original_texts.isna()
            & cleaned_df["text"].isna()
        )
    )

    changed_count = int(changed_mask.sum())
    unchanged_count = len(cleaned_df) - changed_count

    stats = {
        "dataset": dataset_name,
        "total_records": len(cleaned_df),
        "changed_texts": changed_count,
        "unchanged_texts": unchanged_count,
        "missing_texts": int(cleaned_df["text"].isna().sum()),
    }

    return cleaned_df, stats


# Process each dataset separately
df_erzen_cleaned, erzen_stats = process_dataset(
    df_erzen,
    "Erzen"
)

df_arabic_english_cleaned, arabic_english_stats = process_dataset(
    df_arabic_english,
    "Arabic-English"
)

processing_stats = pd.DataFrame([
    erzen_stats,
    arabic_english_stats
])

display(processing_stats)
```

```python
def validate_cleaned_dataset(original, cleaned, name):

    assert len(original) == len(cleaned), (
        f"{name}: Row count changed!"
    )

    assert list(original.columns) == list(cleaned.columns), (
        f"{name}: Columns changed!"
    )

    # Verify every non-text column
    for column in original.columns:

        if column == "text":
            continue

        pd.testing.assert_series_equal(
            original[column],
            cleaned[column],
            check_names=True,
        )

    # Verify normalization idempotency
    normalized_again = cleaned["text"].apply(
        lambda text: normalize_text(text)
        if isinstance(text, str)
        else text
    )

    pd.testing.assert_series_equal(
        cleaned["text"],
        normalized_again,
        check_names=True,
    )

    print(f"PASS | {name}")
    print("  Row count preserved")
    print("  Columns preserved")
    print("  Non-text metadata preserved")
    print("  Normalization is idempotent")


validate_cleaned_dataset(
    df_erzen,
    df_erzen_cleaned,
    "Erzen"
)

validate_cleaned_dataset(
    df_arabic_english,
    df_arabic_english_cleaned,
    "Arabic-English"
)

print("\nDATASET VALIDATION COMPLETED.")
```

```python

# ============================================================
# AI-DS-02 — UNIFIED METADATA SCHEMA
# ============================================================

COMMON_COLUMNS = [
    "utterance_id",
    "audio_path",
    "text",
    "split",
    "speaker_id",
    "recording_id",
    "start",
    "end",
    "duration",
    "source_dataset",
]


# ------------------------------------------------------------
# 1. Remove failed Erzen records from final metadata
# ------------------------------------------------------------

erzen_success_mask = (
    df_erzen["status"]
    .astype("string")
    .str.strip()
    .str.lower()
    .eq("success")
    .fillna(False)
)

erzen_final = df_erzen_cleaned.loc[
    erzen_success_mask
].copy()


# ------------------------------------------------------------
# 2. Standardize Erzen columns
# ------------------------------------------------------------

erzen_final = erzen_final.rename(
    columns={
        "utt_id": "utterance_id",
    }
)

erzen_final["source_dataset"] = "Erzen"

# Keep original utterance IDs, with source prefix
erzen_final["utterance_id"] = (
    "erzen_" + erzen_final["utterance_id"].astype(str)
)


# ------------------------------------------------------------
# 3. Standardize Arabic-English columns
# ------------------------------------------------------------

arabic_final = df_arabic_english_cleaned.copy()

arabic_final = arabic_final.rename(
    columns={
        "meeting": "recording_id",
        "speaker": "speaker_id",
    }
)

# Generate unique IDs using original row indices
arabic_final["utterance_id"] = (
    "arabic_english_"
    + arabic_final.index.astype(str).str.zfill(6)
)

# This dataset does not have an original split column
arabic_final["split"] = "unknown"

arabic_final["source_dataset"] = "Arabic-English"


# ------------------------------------------------------------
# 4. Select the exact same columns and column order
# ------------------------------------------------------------

erzen_final = erzen_final[COMMON_COLUMNS].copy()

arabic_final = arabic_final[COMMON_COLUMNS].copy()


# ------------------------------------------------------------
# 5. Validate the unified schemas
# ------------------------------------------------------------

assert list(erzen_final.columns) == COMMON_COLUMNS
assert list(arabic_final.columns) == COMMON_COLUMNS

assert erzen_final["utterance_id"].is_unique
assert arabic_final["utterance_id"].is_unique

assert erzen_final["source_dataset"].eq("Erzen").all()
assert arabic_final["source_dataset"].eq("Arabic-English").all()


# ------------------------------------------------------------
# 6. Print final metadata statistics
# ------------------------------------------------------------

print("=" * 65)
print("UNIFIED METADATA SCHEMA")
print("=" * 65)

print("\nErzen:")
print("Records:", f"{len(erzen_final):,}")
print("Columns:", len(erzen_final.columns))

print("\nArabic-English:")
print("Records:", f"{len(arabic_final):,}")
print("Columns:", len(arabic_final.columns))

print("\nTotal final records:")
print(f"{len(erzen_final) + len(arabic_final):,}")

print("\nUnified columns:")
print(COMMON_COLUMNS)

print("\nSchema validation passed.")

display(erzen_final.head())
display(arabic_final.head())
```
