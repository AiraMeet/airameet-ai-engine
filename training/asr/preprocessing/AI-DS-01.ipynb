```python
# Read metadata files without modifying the originals
arzen_df = pd.read_csv(ARZEN_METADATA, encoding="utf-8-sig")

arabic_english_df = pd.read_excel(ARABIC_ENGLISH_METADATA)

# Reusable inspection helper (we will move it into src later)
def inspect_metadata(df, dataset_name):
    """
    Display the structure and basic quality information
    of a dataset metadata DataFrame.
    """

    print("=" * 60)
    print(f"Dataset: {dataset_name}")
    print("=" * 60)

    # Dataset dimensions
    print(f"\nRows: {df.shape[0]}")
    print(f"Columns: {df.shape[1]}")

    # Column names
    print("\nColumn names:")
    print(df.columns.tolist())

    # Data types
    print("\nData types:")
    print(df.dtypes)

    # Missing values
    print("\nMissing values per column:")
    print(df.isna().sum())

    # Display a small sample without truncating the dataset
    print("\nFirst 3 rows:")
    display(df.head(3))


# Inspect both datasets
inspect_metadata(arzen_df, "ArZen")
inspect_metadata(arabic_english_df, "Arabic-English-ASR")
```

```python
def audit_metadata(df, dataset_name):
    """
    Inspect metadata for missing values, duplicates,
    invalid time values, and duration inconsistencies.
    """

    print("\n" + "=" * 65)
    print(f"METADATA AUDIT: {dataset_name}")
    print("=" * 65)

    # 1. Missing values
    print("\n[1] Missing values:")
    print(df.isna().sum().to_string())

    # 2. Empty text fields
    print("\n[2] Empty strings:")
    for column in ["audio_path", "text"]:
        if column in df.columns:
            empty_count = (
                df[column]
                .astype("string")
                .str.strip()
                .eq("")
                .sum()
            )
            print(f"{column}: {empty_count}")

    # 3. Duplicate records
    print("\n[3] Duplicate records:")
    print("Duplicate rows:", df.duplicated().sum())

    if "audio_path" in df.columns:
        print(
            "Duplicate audio paths:",
            df["audio_path"].duplicated().sum()
        )

    # 4. Numeric validation
    print("\n[4] Numeric field validation:")

    numeric_columns = ["start", "end", "duration"]

    for column in numeric_columns:
        if column in df.columns:
            values = pd.to_numeric(df[column], errors="coerce")

            invalid_count = (
                values.isna().sum()
                + np.isinf(values.dropna()).sum()
            )

            negative_count = (values < 0).sum()

            print(
                f"{column}: invalid={invalid_count}, "
                f"negative={negative_count}"
            )

    # 5. Duration consistency
    if all(col in df.columns for col in ["start", "end", "duration"]):
        start = pd.to_numeric(df["start"], errors="coerce")
        end = pd.to_numeric(df["end"], errors="coerce")
        duration = pd.to_numeric(df["duration"], errors="coerce")

        calculated_duration = end - start

        valid = (
            start.notna()
            & end.notna()
            & duration.notna()
            & np.isfinite(start)
            & np.isfinite(end)
            & np.isfinite(duration)
        )

        mismatch = (
            (calculated_duration - duration).abs() > 0.05
        ) & valid

        print("\n[5] Duration consistency:")
        print("Rows checked:", valid.sum())
        print("Duration mismatches:", mismatch.sum())


# Run audit on both datasets
audit_metadata(arzen_df, "ArZen")
audit_metadata(arabic_english_df, "Arabic-English-ASR")
```

```python
def metadata_quality_summary(df, dataset_name):
    summary = {
        "Dataset": dataset_name,
        "Rows": len(df),
        "Missing Values": int(df.isna().sum().sum()),
        "Duplicate Rows": int(df.duplicated().sum()),
        "Duplicate Audio Paths": (
            int(df["audio_path"].duplicated().sum())
            if "audio_path" in df.columns else None
        ),
    }

    for column in ["start", "end", "duration"]:
        if column in df.columns:
            values = pd.to_numeric(df[column], errors="coerce")

            summary[f"{column} Invalid"] = int(
                (values.isna() | ~np.isfinite(values)).sum()
            )

            summary[f"{column} Negative"] = int(
                (values < 0).sum()
            )

    if all(c in df.columns for c in ["start", "end", "duration"]):
        start = pd.to_numeric(df["start"], errors="coerce")
        end = pd.to_numeric(df["end"], errors="coerce")
        duration = pd.to_numeric(df["duration"], errors="coerce")

        valid = (
            np.isfinite(start)
            & np.isfinite(end)
            & np.isfinite(duration)
        )

        summary["Duration Mismatches"] = int(
            (((end - start - duration).abs() > 0.05) & valid).sum()
        )

    return summary


# Create a summary for both datasets
quality_report = pd.DataFrame([
    metadata_quality_summary(arzen_df, "ArZen"),
    metadata_quality_summary(arabic_english_df, "Arabic-English-ASR"),
])

display(quality_report.T)
```

```python
def inspect_audio_sample(df, audio_root, sample_size=5):
    results = []

    checked = 0

    for _, row in df.iterrows():
        if checked >= sample_size:
            break

        audio_path = Path(audio_root) / row["audio_path"]

        # Skip missing files
        if not audio_path.is_file():
            continue

        checked += 1

        try:
            info = sf.info(str(audio_path))

            results.append({
                "audio_path": row["audio_path"],
                "sample_rate": info.samplerate,
                "channels": info.channels,
                "duration": round(info.duration, 3),
                "format": info.format,
                "subtype": info.subtype,
                "status": "readable"
            })

        except Exception as error:
            results.append({
                "audio_path": row["audio_path"],
                "status": "decoding_error",
                "error": str(error)
            })

    return pd.DataFrame(results)


arzen_audio_sample = inspect_audio_sample(
    arzen_df,
    ARZEN_ROOT,
    sample_size=5
)

display(arzen_audio_sample)
```

```python
# Inspect signal quality of 5 ArZen audio files

import numpy as np

def inspect_audio_quality(df, audio_root, sample_size=5):
    results = []

    for _, row in df.head(sample_size).iterrows():
        audio_path = Path(audio_root) / row["audio_path"]

        if not audio_path.is_file():
            continue

        try:
            audio, sample_rate = sf.read(
                str(audio_path),
                dtype="float32"
            )

            results.append({
                "audio_path": row["audio_path"],
                "has_nan": bool(np.isnan(audio).any()),
                "has_inf": bool(np.isinf(audio).any()),
                "peak": float(np.max(np.abs(audio))),
                "is_silent": bool(np.max(np.abs(audio)) < 1e-6),
                "clipping_samples": int(
                    np.sum(np.abs(audio) > 0.99)
                ),
            })

        except Exception as error:
            results.append({
                "audio_path": row["audio_path"],
                "error": str(error)
            })

    return pd.DataFrame(results)


arzen_quality_sample = inspect_audio_quality(
    arzen_df,
    ARZEN_ROOT,
    sample_size=5
)

display(arzen_quality_sample)
```

```python
# Calculate clipping percentage for all ArZen audio files

clipping_stats = arzen_quality_audit.copy()

# Get total audio frames for each file
clipping_stats["total_samples"] = clipping_stats["audio_path"].apply(
    lambda path: sf.info(str(ARZEN_ROOT / path)).frames
)

# Calculate clipping percentage
clipping_stats["clipping_percentage"] = (
    clipping_stats["clipping_samples"]
    / clipping_stats["total_samples"]
    * 100
)

print("Total files:", len(clipping_stats))
print(
    "Files with clipping:",
    (clipping_stats["clipping_samples"] > 0).sum()
)

print("\nClipping percentage statistics:")
display(clipping_stats["clipping_percentage"].describe())

print("\nTop 10 files with highest clipping:")
display(
    clipping_stats.sort_values(
        "clipping_percentage",
        ascending=False
    ).head(10)
)
```

```python
# Classify clipping severity for investigation

def classify_clipping(value):
    if value == 0:
        return "No Clipping"
    elif value <= 0.1:
        return "Very Low"
    elif value <= 1.0:
        return "Moderate"
    else:
        return "High"

clipping_stats["clipping_level"] = (
    clipping_stats["clipping_percentage"]
    .apply(classify_clipping)
)

print("Clipping severity distribution:")
display(
    clipping_stats["clipping_level"]
    .value_counts()
    .to_frame("File Count")
)

print("\nAverage clipping percentage by level:")
display(
    clipping_stats.groupby("clipping_level")[
        "clipping_percentage"
    ].mean().to_frame("Average Clipping %")
)
```

```python
# Inspect high-clipping audio files

high_clipping_files = clipping_stats[
    clipping_stats["clipping_level"] == "High"
].sort_values(
    "clipping_percentage",
    ascending=False
)

display(
    high_clipping_files[
        [
            "audio_path",
            "peak",
            "clipping_samples",
            "total_samples",
            "clipping_percentage"
        ]
    ]
)

print("High clipping files:", len(high_clipping_files))
```

```python
# Build the final audio manifest

from pathlib import Path

final_audio_manifest = arabic_metadata_df.copy()

def resolve_audio_path(relative_path):
    relative_path = Path(str(relative_path))

    processed_path = processed_root / relative_path
    raw_path = raw_root / relative_path

    # Prefer the standardized processed file
    if processed_path.is_file():
        return str(processed_path), "processed"

    # Otherwise use the original file
    if raw_path.is_file():
        return str(raw_path), "raw"

    return None, "missing"


resolved_paths = final_audio_manifest["audio_path"].apply(
    resolve_audio_path
)

final_audio_manifest["final_audio_path"] = (
    resolved_paths.apply(lambda item: item[0])
)

final_audio_manifest["audio_source"] = (
    resolved_paths.apply(lambda item: item[1])
)

# Display summary
print("Total metadata records:", len(final_audio_manifest))

print("\nAudio source distribution:")
display(final_audio_manifest["audio_source"].value_counts())

print(
    "\nMissing audio files:",
    final_audio_manifest["final_audio_path"].isna().sum()
)

display(
    final_audio_manifest[
        ["audio_path", "final_audio_path", "audio_source"]
    ].head(10)
)
```

```python
from tqdm.auto import tqdm

audio_quality_results = []

for _, row in tqdm(final_audio_manifest.iterrows(),
                   total=len(final_audio_manifest)):

    audio_path = Path(row["final_audio_path"])

    result = {
        "audio_path": str(audio_path),
        "readable": False,
        "has_nan_inf": None,
        "peak_amplitude": None,
        "is_clipped": None,
        "error": None,
    }

    try:
        audio, sample_rate = sf.read(
            audio_path,
            dtype="float32"
        )

        result["readable"] = True
        result["has_nan_inf"] = not np.isfinite(audio).all()

        if not result["has_nan_inf"]:
            peak = float(np.max(np.abs(audio)))
            result["peak_amplitude"] = peak
            result["is_clipped"] = peak > 0.99

    except Exception as error:
        result["error"] = str(error)

    audio_quality_results.append(result)

audio_quality_df = pd.DataFrame(audio_quality_results)

print("Total checked:", len(audio_quality_df))
print("\nReadable files:")
display(audio_quality_df["readable"].value_counts())

print("\nNaN / Inf:")
display(audio_quality_df["has_nan_inf"].value_counts(dropna=False))

print("\nClipping:")
display(audio_quality_df["is_clipped"].value_counts(dropna=False))
```

```python
snr_results = []

for _, row in tqdm(
    final_audio_manifest.iterrows(),
    total=len(final_audio_manifest),
    desc="Calculating WADA-SNR"
):
    audio_path = Path(row["final_audio_path"])

    result = {
        "audio_path": str(audio_path),
        "snr_db": None,
        "status": "failed",
        "error": None
    }

    try:
        # Read audio as a NumPy array
        audio, sample_rate = sf.read(
            audio_path,
            dtype="float32"
        )

        # Ensure mono audio
        if audio.ndim == 2:
            audio = audio.mean(axis=1)

        # Calculate WADA-SNR
        snr_value = wada_snr(audio)

        result["snr_db"] = snr_value
        result["status"] = "success"

    except Exception as e:
        result["error"] = str(e)

    snr_results.append(result)


# Convert results to DataFrame
snr_df = pd.DataFrame(snr_results)

print("Total files:", len(snr_df))
print("\nProcessing status:")
display(snr_df["status"].value_counts())

print("\nSNR statistics:")
display(snr_df["snr_db"].describe())

print("\nFailed files:")
display(
    snr_df[snr_df["status"] == "failed"].head(20)
)
```

```python

# Complete handoff directory
handoff_audio_dir = Path(
    "data/processed/Arabic-English-ASR/complete_audio"
)

handoff_audio_dir.mkdir(parents=True, exist_ok=True)

# Copy all files using the relative audio paths
copied_files = []
missing_files = []

for _, row in final_audio_manifest.iterrows():

    source_path = Path(row["final_audio_path"])
    relative_path = Path(row["audio_path"])

    destination_path = handoff_audio_dir / relative_path

    if not source_path.exists():
        missing_files.append(str(relative_path))
        continue

    destination_path.parent.mkdir(parents=True, exist_ok=True)

    shutil.copy2(source_path, destination_path)

    copied_files.append(str(relative_path))

print("Copied files:", len(copied_files))
print("Missing files:", len(missing_files))
print("Handoff directory:", handoff_audio_dir)
```
